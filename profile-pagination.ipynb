```python
from pathlib import Path

import pandas as pd

frames = []

for path in Path.cwd().glob("1*.json"):
    df = pd.read_json(path)
    df["ref"] = path.stem.strip()
    frames.append(df)

df = pd.concat(frames)
df
```

```python
import altair as alt
```

```python
(alt.Chart(df)
        .mark_line()
        .encode(
            x="n",
            y=alt.Y("t", title="t (seconds)"),
            color="per_page:N",
            # row="per_page",
            column="api",
        )
        .properties(width=400, height=180)
)
```

```python
alt.vconcat(
    *[
        alt.Chart(df[df.per_page == p], title=f"{p} per page")
        .mark_line()
        .encode(
            x="n",
            y=alt.Y("t", title="t (seconds)"),
            color="ref",
            # row="per_page",
            column="api",
        )
        .properties(width=400, height=180)
        for p in sorted(df.per_page.unique())
    ]
)
```

```python
df["requests"] = df.n / df.per_page
df.loc[df.requests < 1, "requests"] = 1
```

```python
df["requests/sec"] = df["requests"] / df.t
df["obs/sec"] = df.n / df.t
```

```python
(alt.Chart(df)
        .mark_point()
        .encode(
            x="per_page:Q",
            y=alt.Y("requests/sec", title="requests per second"),
            color="n:Q",
            # row="per_page",
            column="api",
        )
        #.properties(width=400, height=180)
)
```

```python
(alt.Chart(df)
        .mark_point()
        .encode(
            x="per_page:Q",
            y=alt.Y("obs/sec", title="observations per second"),
            color="n:Q",
            # row="per_page",
            column="api",
        )
)
```
